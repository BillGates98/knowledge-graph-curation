```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv('./test_data/data.csv')

df['NumPerm-Threshold'] = df['NumPerm-Threshold'].apply(lambda x: str(x))

df.drop(columns=['KGs', 'Axis'], inplace=True)
df_grouped = df.groupby('NumPerm-Threshold').mean().reset_index()

# df = pd.DataFrame([['A', 10, 20, 10, 30], ['B', 20, 25, 15, 25], ['C', 12, 15, 19, 6],
#                    ['D', 10, 29, 13, 19]],
#                   columns=['Team', 'Round 1', 'Round 2', 'Round 3', 'Round 4'])

# plot grouped bar chart
df_grouped.plot(x='NumPerm-Threshold',
        kind='bar',
        stacked=False,
        title='Runtime vs Number-of-Permutations - Threshold',
        xlabel='Number-of-Permutations - Threshold',
        ylabel='Runtime (seconds)',
        figsize=(10, 6))
plt.xticks(rotation=0)
plt.show()
plt.savefig('runtime_vs_numperm_threshold.png')
```
